# Data Inspection - Combined Cycle Power Plant

## Sumber Dataset

Dataset yang digunakan adalah Combined Cycle Power Plant (CCPP).

Menurut dokumentasi sumber, dataset terdiri dari 9.568 data point yang
dikumpulkan dari sebuah Combined Cycle Power Plant selama 6 tahun,
yaitu 2006–2011.

Dataset menyediakan lima sheet yang merupakan hasil pengacakan data
untuk kebutuhan 5x2-fold statistical testing. Karena kelima sheet
merepresentasikan data yang sama dalam urutan berbeda, penelitian ini
menggunakan Sheet1 sebagai sumber data utama dan tidak menggabungkan
kelima sheet.

## Variabel

| Kolom | Keterangan |
|---|---|
| AT | Ambient Temperature |
| V | Exhaust Vacuum |
| AP | Ambient Pressure |
| RH | Relative Humidity |
| PE | Electrical Power Output |

## Kondisi Awal Dataset

- Jumlah observasi: 9.568
- Jumlah variabel: 5
- Missing value: tidak ditemukan
- Duplikasi identik: 41 baris
- Seluruh variabel bertipe numerik
