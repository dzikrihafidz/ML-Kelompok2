```python
import pandas as pd
import numpy as np

file_path = "../data/raw/Folds5x2_pp.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="Sheet1"
)

df.head()
```

```python
print("Shape awal:", df.shape)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate:")
print(df.duplicated().sum())

print("\nInfinite:")
print(np.isinf(df).sum().sum())
```

```python
before = len(df)

df_clean = df.drop_duplicates().copy()

after = len(df_clean)

print("Jumlah baris sebelum:", before)
print("Jumlah baris sesudah:", after)
print("Jumlah duplicate dihapus:", before - after)
```

## Penanganan Outlier

Berdasarkan hasil EDA, terdapat kandidat outlier secara statistik pada
beberapa variabel, terutama AP dan RH.

Namun, kandidat outlier tidak langsung dihapus karena belum terdapat
bukti bahwa nilai tersebut merupakan kesalahan pengukuran atau data
invalid. Nilai ekstrem masih berada dalam rentang nilai yang terdapat
pada dataset asli dan dapat merepresentasikan variasi kondisi operasi.

Oleh karena itu, tahap cleaning hanya menghapus baris duplicate identik.
Kandidat outlier dipertahankan untuk digunakan pada tahap pemodelan.

```python
print("Missing values:")
print(df_clean.isnull().sum())

print("\nDuplicate:")
print(df_clean.duplicated().sum())

print("\nInfinite:")
print(np.isinf(df_clean).sum().sum())
```

```python
print(df_clean.columns.tolist())
```

```python
output_path = "../data/processed/CCPP_clean.csv"

df_clean.to_csv(
    output_path,
    index=False
)

print("Dataset clean berhasil disimpan:")
print(output_path)
```

## Kesimpulan Data Cleaning

Dataset awal terdiri dari 9.568 observasi dan 5 variabel.

Hasil validasi menunjukkan tidak terdapat missing value maupun nilai
infinite. Terdapat 41 baris duplicate identik yang kemudian dihapus
karena tidak memberikan informasi baru.

Setelah proses cleaning, dataset terdiri dari 9.527 observasi dan

operasi pada dataset.
